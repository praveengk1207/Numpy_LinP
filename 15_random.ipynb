Random In python Numpy


From this we use "Random module" which is imported from numpy 


```python
#To import random in code
from numpy import random
```

We can generate 

--> Random Integer. 

--> Random Float values.

--> Random of our choice.

--> Random in the form of multi dimensional arrays / nd array.


----------------------------------------------------------------------------

For Random integers we use randint(range).

Eg: num = random.randint(100)

op: num b/w (0 to 100)

Note: every thime we execute the code we get new random integer

```python
from numpy import random
num = random.randint(100)
print(f'The random integer: {num}')
```

---

For Random Float values we use rand(size)

Eg num = random.rand(2)

op: [0.121 0.333]

Note it returns values in array format when you pass size as a argument else it just returns random number as a output

```python
from numpy import random
num = random.rand()
print(f'float values: {num}')
```

```python
from numpy import random
num = random.rand(5)
print(f"array of float values: {num}")
```

---

From Random of our choice we use choice([array of values separated by commas])

eg: num = random.choice([8,5,6,7])

op: 6



```python
from numpy import random
num = random.choice([8,5,6,9,2])
print(f'random choice value: {num}')
```

---

For creating random multi dimensional array 

multi dim with integers or randint()

```python
# with integers
# we use size for it 
# size = (dim,no_of_ele)
#eg:

from numpy import random
num = random.randint(100, size=(3,3))
print(num)
```

multi dim with float values or rand()

```python
# with float values
# we can directly define dim as first argument and as we already know we alredy have no_of_ele
#eg rand(dim,no_of_ele)

from numpy import random
num = random.rand(3,5)
print(num)
```

multi dim with random.choice()

```python
# with choice()
#eg random.choice([array of ele], size = (dim,no_of_ele))


from numpy import random
num = random.choice([8,5,4,2,1], size=(3,5))
print(num)
```
