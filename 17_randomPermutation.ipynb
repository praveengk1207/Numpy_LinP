```python
from numpy import random
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
random.shuffle(arr)
print(arr)
```

```python
from numpy import random
import numpy as np
arr = np.array([1, 2, 3, 6, 5])
arr2 = random.permutation(arr)
print(arr2)
```

Note :
when we use permutation method it doesnt makes any changes to original array insted it returns a re-arranged array
